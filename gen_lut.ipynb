```python
def generate_verilog_lut(input_txt, output_v, module_name="range_exp_lut"):
    try:
        with open(input_txt, 'r') as f:
            lines = f.readlines()

        with open(output_v, 'w') as out:
            # 寫入 Verilog 模組標頭
            out.write(f"module {module_name} (\n")
            out.write("    input  wire [13:0] addr,\n")
            out.write("    output reg  [10:0] data\n")
            out.write(");\n\n")
            out.write("always @(*) begin\n")
            out.write("    case(addr)\n")

            # 處理每一行資料
            for line in lines:
                line = line.strip()
                if not line: continue  # 跳過空行
                
                parts = line.split()
                if len(parts) == 2:
                    idx = int(parts[0])
                    val = int(parts[1])
                    # 格式化為 Verilog 的數值表示法: 14'd0: data = 11'd2047;
                    out.write(f"        14'd{idx:<5}: data = 11'd{val};\n")

            # 結束 case 與 module
            out.write("        default: data = 11'd0;\n")
            out.write("    endcase\n")
            out.write("end\n\n")
            out.write("endmodule\n")
        
        print(f"成功生成 Verilog 檔案: {output_v}")

    except FileNotFoundError:
        print("錯誤: 找不到輸入的 txt 檔案")
    except ValueError:
        print("錯誤: txt 內容格式不正確，請確保 index 和 value 為整數")

# 執行轉換
generate_verilog_lut("./lut/My_range_exp_LUT.txt", "./rtl_lut_file/range_exp_lut.v")
```
